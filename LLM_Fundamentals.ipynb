## What is a Large Language Model?

A Large Language Model (LLM) is a neural network trained on massive amounts of text to predict the next token (word or subword) in a sequence.[8]  
Because it learns patterns in language, it can generate text, answer questions, summarize, translate, and act like a conversational assistant.[8]

**Simple diagram (conceptual):**

- Input text → tokenization → model processes tokens → output tokens → decoded back to text

## LLMs vs Traditional ML

Traditional ML models usually learn a mapping from features to labels, for example spam vs not spam or fraud vs not fraud.

LLMs are different because they are trained on language itself and generate output one token at a time.

**Key difference:**
- Traditional ML: fixed input features → prediction
- LLMs: text tokens → next-token prediction → generated text

## Tokens and Tokenization

LLMs don’t work directly on raw characters or words; they work on tokens, which are pieces of text like words or subwords.[15]  
Tokenization is the process of converting raw text into a sequence of tokens that the model can understand.[15]

**Example idea:**
- Text: `"Large Language Models are changing AI."`
- Tokens: `["large", "language", "models", "are", "changing", "ai", "."]` (exact tokens depend on the tokenizer)

```python

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "Large Language Models are changing AI."

tokens = tokenizer.tokenize(text)
print("Tokens:", tokens)
print("Number of tokens:", len(tokens))
```

## Context Window

An LLM can only see a limited number of tokens at once; this limit is called the context window.[13]  
If your prompt plus conversation history exceed this limit, older tokens are truncated or ignored, which is why long chats sometimes “forget” earlier details.[13]

**Intuition:**
- Small context window → short memory
- Large context window → can consider more instructions, documents, or conversation history at once

## Embeddings

Embeddings are numerical representations of text that capture meaning in a vector form.

Words with similar meanings often have embeddings that are close together in vector space.

You can think of embeddings as a way for models to “understand” semantic similarity, not just exact words.

## Transformer Architecture

Transformers are the core architecture behind modern LLMs.

At a high level, transformers use:
- Self-attention to focus on important words in context.
- Layers stacked together to learn complex language patterns.

You do not need to learn the full math today — just understand that transformers help LLMs pay attention to relationships between tokens.

## Pretraining vs Fine-tuning

Pretraining is when an LLM learns general language patterns from a massive amount of text.

Fine-tuning is when that pretrained model is adapted for a specific task or domain.

**Simple idea:**
- Pretraining = learn the language
- Fine-tuning = specialize for a job

## Inference

Inference is the stage where a trained LLM generates output for your prompt.

At inference time, the model:
- Reads your input prompt.
- Predicts the next token.
- Repeats that process until it finishes the response.

**Example:**  
You type a question, and the model generates an answer one token at a time.

## Temperature, Top-k, and Top-p

These are sampling settings that control how an LLM chooses the next token.

- **Temperature** controls randomness. Lower values make output more focused; higher values make it more creative.
- **Top-k** limits choices to the top \(k\) most likely tokens.
- **Top-p** (nucleus sampling) keeps the smallest set of tokens whose total probability reaches \(p\).

**Simple idea:**
- Low temperature = safer, more deterministic
- High temperature = more creative, less predictable

## Hallucinations

Hallucinations happen when an LLM produces text that sounds confident but is factually wrong.

This can happen because the model is generating the most likely next tokens, not checking facts from a database or the real world.

**Why it matters:**
- The response may sound correct.
- But it can include made-up details, wrong dates, or false references.

## Prompt Engineering Basics

Prompt engineering is the practice of writing clear instructions so the LLM gives better output.

Good prompts usually include:
- The task.
- The context.
- The format you want.
- Any constraints or examples.

**Example:**  
Instead of saying “Explain tokenization,” say:  
“Explain tokenization in simple words with one short example.”

## Simple Diagram

Text → Tokenizer → Tokens → LLM → Output

**Flow:**
- Raw text goes in.
- Tokenizer splits it into tokens.
- LLM processes the tokens.
- The model generates output token by token.

## Key Takeaways

- LLMs generate text one token at a time.
- Tokenization converts text into model-readable tokens.
- Context windows limit how much information a model can consider.
- Embeddings capture semantic meaning rather than exact words.
- Transformers use self-attention to understand relationships between words.
- Pretraining teaches general language patterns, while fine-tuning specializes a model.
- Prompt engineering significantly influences the quality of responses.
- Temperature, Top-k, and Top-p control response creativity.
